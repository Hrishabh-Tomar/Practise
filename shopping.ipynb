```python
cart = ["Laptop", "Mouse", "Keyboard", "Monitor", "Headphones"]

# Display all products
print("All products:", cart)

# Access first and last products
print("First product:", cart[0])
print("Last product:", cart[-1])

# Add "Webcam"
cart.append("Webcam")
print("\nAfter append('Webcam'):", cart)

# Insert "USB Hub" at index 2
cart.insert(2, "USB Hub")
print("After insert(2, 'USB Hub'):", cart)

# Remove "Mouse"
cart.remove("Mouse")
print("After remove('Mouse'):", cart)

# Remove the last item using pop()
popped_item = cart.pop()
print(f"After pop() (removed '{popped_item}'):", cart)

# Find the index of "Monitor"
monitor_index = cart.index("Monitor")
print("\nIndex of 'Monitor':", monitor_index)

# Count occurrences of "Laptop"
laptop_count = cart.count("Laptop")
print("Count of 'Laptop':", laptop_count)

# Create a copy of the cart
cart_copy = cart.copy()
print("\nCopy of cart:", cart_copy)

# Reverse the cart
cart.reverse()
print("After reverse():", cart)

# Sort the products alphabetically
cart.sort()
print("After sort():", cart)
```

```python
# Add multiple products at once using extend()
cart.extend(["Charger", "Mouse Pad"])
print("After extend(['Charger', 'Mouse Pad']):", cart)
```

```python
# Clear the entire cart
cart_backup = cart.copy()  # save a copy first so we still have data to work with
cart.clear()
print("After clear():", cart)

# Restore for the rest of the notebook
cart = cart_backup
```

List Method Explanations
append(x) — adds a single item x to the end of the list.
extend(iterable) — adds each item from another list (or iterable) individually to the end of the list, instead of adding it as one nested element.
insert(i, x) — inserts item x at a specific index i, shifting later elements to the right.
remove(x) — removes the first occurrence of value x from the list (raises an error if not found).
pop(i) — removes and returns the item at index i (defaults to the last item if no index is given).
clear() — removes all items from the list, leaving it empty ([]).
copy() — returns a shallow copy of the list, so changes to the copy don't affect the original.
sort() — sorts the list in place in ascending order (or by a custom key, if provided).
reverse() — reverses the order of elements in the list in place.
