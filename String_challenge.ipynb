## String Challenge

```python
# Starting message with leading/trailing whitespace
message = "   Welcome To Python Programming Class   "  # placeholder — replace once exact spec is confirmed

# 1. Remove extra spaces
stripped_message = message.strip()
print("original:", repr(message))
print("stripped:", repr(stripped_message))

# 2. Convert to lower case
lower_case = stripped_message.lower()
print(lower_case)

# 3. Convert to upper case
upper_message = stripped_message.upper()
print(upper_message)

# 4. Convert to title case
title_message = stripped_message.title()
print(title_message)

# 5. Replace "Python" with "Advanced Python"
replaced_message = stripped_message.replace("Python", "Advanced Python")
print("\nAfter replace('Python', 'Advanced Python'):", replaced_message)

# 6. Check whether it starts with "Welcome"
starts_check = stripped_message.startswith("Welcome")
print("\nStarts with 'Welcome':", starts_check)

# 7. Check whether it ends with "Class"
ends_check = stripped_message.endswith("Class")
print("Ends with 'Class':", ends_check)

# 8. Count occurrences of "o"
o_count = stripped_message.count("o")
print("\nCount of 'o':", o_count)

# 9. Find the position of "Programming"
position = stripped_message.find("Programming")
print("Position of 'Programming':", position)

# 10. Split the sentence into words
words = stripped_message.split()
print("\nSplit into words:", words)
```

String Method Explanations
strip() — removes leading and trailing whitespace (spaces, tabs, newlines) from the string; it does not affect spaces in the middle of the text.
lower() — returns a new string with every character converted to lowercase.
upper() — returns a new string with every character converted to uppercase.
title() — returns a new string where the first letter of each word is capitalized and the rest are lowercased (e.g., "welcome to python" → "Welcome To Python").
replace(old, new) — returns a new string with all occurrences of old replaced by new; the original string stays unchanged because strings are immutable in Python.
startswith(prefix) / endswith(suffix) — return True/False depending on whether the string begins or ends with the given substring, without needing manual slicing.
count(sub) — returns how many non-overlapping times sub appears in the string.
find(sub) — returns the index of the first occurrence of sub, or -1 if not found (unlike .index(), which raises an error on failure).
split(sep=None) — breaks the string into a list of substrings on the given delimiter; with no argument, it splits on any whitespace and ignores extra/multiple spaces automatically.
