```python
python_students = {"Rahul", "Aman", "Priya", "Karan", "Neha"}
java_students = {"Priya", "Karan", "Rohit", "Simran"}

# Print both sets
print("Python students:", python_students)
print("Java students:", java_students)

# Find students learning either Python or Java (union)
either_students = python_students.union(java_students)
print("\nStudents learning either Python or Java (union):", either_students)

# Find students learning both (intersection)
both_students = python_students.intersection(java_students)
print("Students learning both (intersection):", both_students)

# Find students learning only Python (difference)
only_python = python_students.difference(java_students)
print("Students learning only Python (difference):", only_python)

# Find students learning only Java (difference)
only_java = java_students.difference(python_students)
print("Students learning only Java (difference):", only_java)

# Find students belonging to exactly one group (symmetric_difference)
exactly_one_group = python_students.symmetric_difference(java_students)
print("\nStudents in exactly one group (symmetric_difference):", exactly_one_group)

# Add a new student
python_students.add("Divya")
print("\nAfter add('Divya') to python_students:", python_students)

# Remove a student
python_students.remove("Rahul")
print("After remove('Rahul') from python_students:", python_students)

# Demonstrate discard()
python_students.discard("Rahul")  # already removed - discard won't raise an error
print("After discard('Rahul') again (already removed, no error):", python_students)

python_students.discard("Neha")   # exists, so this removes it normally
print("After discard('Neha') (existing member):", python_students)
```
