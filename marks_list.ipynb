```python
marks = [78, 85, 90, 67, 88, 92, 76]

# Print the complete list
print("Complete list:", marks)

# Print the first element
print("\nFirst element:", marks[0])

# Print the last element
print("Last element:", marks[-1])

# Print elements from index 2 to 5
print("\nElements from index 2 to 5:", marks[2:6])

# Find the number of elements
print("\nNumber of elements:", len(marks))

# Find maximum marks
print("Maximum marks:", max(marks))

# Find minimum marks
print("Minimum marks:", min(marks))

# Find total marks
print("Total marks:", sum(marks))

# Sort marks in ascending order
ascending_marks = sorted(marks)
print("\nSorted ascending:", ascending_marks)

# Sort marks in descending order
descending_marks = sorted(marks, reverse=True)
print("Sorted descending:", descending_marks)

# Add 95
marks.append(95)
print("\nAfter adding 95:", marks)

# Add [81, 84]
marks.extend([81, 84])
print("After extending with [81, 84]:", marks)

# Remove 67
marks.remove(67)
print("\nAfter removing 67:", marks)

# Count how many times 90 occurs
print("\nCount of 90:", marks.count(90))

# Find the index of 88
print("Index of 88:", marks.index(88))
```
