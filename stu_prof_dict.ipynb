```python
student = {
    "name": "Rahul",
    "age": 22,
    "course": "Python",
    "city": "Bangalore",
    "marks": 88
}

# Print the complete dictionary
print("Complete dictionary:", student)

# Print the student's name
print("\nName:", student["name"])

# Print their course
print("Course:", student["course"])

# Print all keys
print("\nAll keys:", student.keys())

# Print all values
print("All values:", student.values())

# Print all key-value pairs
print("All key-value pairs:", student.items())

# Change marks from 88 to 92
student["marks"] = 92
print("\nAfter changing marks:", student["marks"])

# Add "email"
student["email"] = "rahul@example.com"
print("After adding email:", student["email"])

# Add "phone"
student["phone"] = "9876543210"
print("After adding phone:", student["phone"])

# Remove "city"
student.pop("city")
print("\nAfter removing 'city':", student)

# Use get() to retrieve "name"
name_value = student.get("name")
print("\nUsing get('name'):", name_value)

# Using get() with a key that doesn't exist
grade_value = student.get("grade", "Not Available")
print("Using get('grade') with default:", grade_value)

# Create a copy of the dictionary
student_copy = student.copy()
print("\nCopy of dictionary:", student_copy)

# Demonstrate update()
student.update({"marks": 95, "city": "Hyderabad"})
print("\nAfter update({'marks': 95, 'city': 'Hyderabad'}):", student)
print("(Original copy remains unaffected:", student_copy, ")")
```

Dictionary Method Explanations
dict[key] — direct indexing to access a value by its key; raises a KeyError if the key doesn't exist. Used here to read student["name"] and student["course"], and also to add or update keys via assignment (student["marks"] = 92, student["email"] = "...") — if the key already exists its value is overwritten, if not, a new key-value pair is created.
.keys() — returns a view object containing all the keys in the dictionary, useful for iterating over or checking what fields exist without touching the values.
.values() — returns a view object containing all the values in the dictionary, in the same order as the keys, without any information about which key each value belongs to.
.items() — returns a view object of (key, value) tuple pairs, letting you access both the field name and its value together (typically used when looping with for k, v in student.items()).
.pop(key) — removes the specified key from the dictionary and returns its value; unlike del dict[key], it lets you capture what was removed, and you can pass a default to avoid an error if the key is missing.
.get(key, default) — safely retrieves a value by key without raising an error if the key doesn't exist; instead it returns None, or the default value you specify (e.g., student.get("grade", "Not Available")). This is safer than dict[key] when you're not sure a key is present.
.copy() — creates a shallow copy of the dictionary — a completely separate dictionary object with the same key-value pairs, so modifying the copy (or the original) afterward doesn't affect the other. This is why student_copy stays unchanged even after student is updated later.
.update(other_dict) — merges another dictionary into this one: existing keys get their values overwritten ("marks" changes from 92 to 95), and new keys get added ("city" is reintroduced with a new value) — all in a single call, instead of assigning each key one at a time.
