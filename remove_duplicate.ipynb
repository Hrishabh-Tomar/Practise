```python
numbers = [10, 20, 30, 20, 40, 10, 50, 30, 60]

# Print the original list
print("Original list:", numbers)

# Convert the list into a set
numbers_set = set(numbers)
print("\nConverted to set:", numbers_set)
print("(Notice duplicate values 10, 20, and 30 each appear only once now)")

# Convert the set back into a list
unique_numbers_list = list(numbers_set)
print("\nConverted back to list:", unique_numbers_list)

# Print the number of original elements
print("\nNumber of original elements:", len(numbers))

# Print the number of unique elements
print("Number of unique elements:", len(numbers_set))

# ------------------------------------------------------------
# Another example: duplicate student names
# ------------------------------------------------------------
student_names = ["Rahul", "Priya", "Aman", "Rahul", "Sneha", "Priya", "Rahul"]
print("\nOriginal list of names:", student_names)

unique_names_set = set(student_names)
print("Converted to set:", unique_names_set)

unique_names_list = list(unique_names_set)
print("Converted back to list:", unique_names_list)

print("\nNumber of original name entries:", len(student_names))
print("Number of unique students:", len(unique_names_set))
```
