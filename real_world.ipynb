Cell 1 — Total transaction value

```python
# total sum
transactions = [1200, 450, 800, 1500, 2300, 700, 100]

total = 0
for amount in transactions:
    total += amount

print(f"Total transactions is :{total}")
```

Cell 2 — Highest & lowest transaction

```python
# Highest and Lowest
transactions = [1200, 450, 800, 1500, 2300, 700, 100]

highest = transactions[0]
lowest  = transactions[0]

for amount in transactions:
    if amount > highest:
        highest = amount
    if amount < lowest:
        lowest = amount
    
print(f"Highest: {highest}, Lowest : {lowest}")
```

Cell 3 — Average temperature

```python
# Average Temp
temperatures = [32, 35, 28, 40, 38, 31, 42]

total = 0
count = 0

for temp in temperatures:
    total += temp
    count += 1

Average_temp = total / count
print(f"Average temperature :{Average_temp}")
```

Cell 4 — Marks distribution

```python
#Count students by score band
marks = [78, 92, 45, 67, 88, 53, 99]

above_90 = 0
betw_89_75 = 0
betw_74_50 = 0
below_50 = 0

for mark in marks:
    if mark >= 90:
        above_90 += 1
    elif  mark >= 75:
        betw_89_75 += 1
    elif mark >= 50:
        betw_74_50 += 1
    else: 
        below_50 += 1
        
print("90+ :",above_90) 
print("89-75 :", betw_89_75)
print("74-50 :", betw_74_50)
print("<50 :", below_50)
```

Cell 5 — Login system (3 attempts)

```python
# Login system (max 3 attempts)
correct_password = "python123"
Max_attempts = 3

attempts = 0
while attempts < Max_attempts:
    typed_password = input("enter your password: ")
    attempts += 1
    if typed_password == correct_password:
        print("login Successful !!")
        break
    else:
        remaining = Max_attempts - attempts
        if remaining > 0:
            print(f"Attempts left {remaining}")
        else:
            print(f"exceeded attempts account locked")

```

Cell 8 — Character frequency (no Counter)

```python
#Character frequency (without Counter)
text = "banana"
freq = {}

for ch in text:
    if ch in freq:
        freq[ch] += 1
    else:
        freq[ch] = 1
    
for ch, count in freq.items():
    print(f"{ch} : {count}")
```

Cell 9 — Second-largest number (no sort())

```python
#Second-largest number (without sort())
numbers = [45, 12, 89, 89, 33, 67, 21]

largest = second = float("-inf")

for num in numbers:
    if num > largest:
        second = largest
        largest = num
    elif largest > num > second:
        second = num
        
print("second largest:", second)
```

Cell 10 — Palindrome check (two-pointer)

```python
#Palindrome check using loops
word = "malayalam"

left = 0
right = len(word) -1
is_palindrome = True

while left < right:
    if word[left] != word[right]:
        is_palindrome = False
        break
    left += 1
    right -= 1

print(f"{word} is {'a palindrome' if is_palindrome else 'not a palindrome'}")
    
```

Cell 12 — ATM simulation (modular)

```python
# ATM Simulation

def show_menu():
    print("\n===== ATM MENU =====")
    print("1. Check Balance")
    print("2. Deposit Money")
    print("3. Withdraw Money")
    print("4. Exit")
```

```python
def check_balance(balance):
    print(f"Current balance: {balance}")
    return balance
```

```python

def deposit(balance):
    amount = float(input("Enter deposit amount: "))
    if amount <= 0:
        print("Error: deposit amount must be positive.")
        return balance
    balance += amount
    print(f"Deposited {amount}. New balance: {balance}")
    return balance
```

```python
def withdraw(balance):
    amount = float(input("Enter withdrawal amount: "))
    if amount <= 0:
        print("Error: withdrawal amount must be positive.")
    elif amount > balance:
        print("Error: insufficient funds.")
    else:
        balance -= amount
        print(f"Withdrew {amount}. New balance: {balance}")
    return balance
```

```python
def run_atm(starting_balance=1000):
    balance = starting_balance
    while True:
        show_menu()
        choice = input("Choose an option (1-4): ").strip()

        if choice == "1":
            balance = check_balance(balance)
        elif choice == "2":
            balance = deposit(balance)
        elif choice == "3":
            balance = withdraw(balance)
        elif choice == "4":
            print("Thank you for using the ATM. Goodbye!")
            break
        else:
            print("Invalid option. Please choose 1-4.")


run_atm()
```
