```python
message = "    Welcome to python programming course     "
```

```python
stripped_msg = message.strip()
print("edited:", stripped_msg)
print("original:", message)
```

```python
lower_msg = stripped_msg.lower()
print(lower_msg)
```

```python
upper_msg = stripped_msg.upper()
print(upper_msg)
```

```python
title_msg = stripped_msg.title()
print(title_msg)
```

```python
replace_msg = stripped_msg.replace("python", "Advanced Python")
print(replace_msg)
```

```python
starts_with = stripped_msg.startswith("Welcome")
print(starts_with)
```

```python
ends_with = stripped_msg.endswith("course")
print(ends_with)
```

```python
count_o = stripped_msg.count("o")
print(count_o)
```

```python
prog_position = stripped_msg.index("programming")
print(prog_position)
```

```python
#spilt into words
strip_words = stripped_msg.split()
print(strip_words)
```
