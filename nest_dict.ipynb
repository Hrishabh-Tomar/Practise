```python
employee = {
    "name": "Amit",
    "department": "Engineering",
    "skills": {
        "language": "Python",
        "database": "PostgreSQL",
        "cloud": "AWS"
    },
    "salary": 80000
}

# Print employee name
print("Employee name:", employee["name"])

# Print department
print("Department:", employee["department"])

# Print complete skills dictionary
print("\nComplete skills dictionary:", employee["skills"])

# Print programming language
print("\nProgramming language:", employee["skills"]["language"])

# Print database
print("Database:", employee["skills"]["database"])

# Print cloud technology
print("Cloud technology:", employee["skills"]["cloud"])

# Change "Python" to "Python + JavaScript"
employee["skills"]["language"] = "Python + JavaScript"
print("\nAfter updating language:", employee["skills"]["language"])

# Change salary
employee["salary"] = 90000
print("After updating salary:", employee["salary"])

# Add "experience": 3
employee["experience"] = 3
print("After adding experience:", employee["experience"])

# Add another skill under the skills dictionary
employee["skills"]["framework"] = "Django"
print("After adding new skill under skills:", employee["skills"])

# Print the fully updated employee dictionary
print("\nFinal updated employee dictionary:", employee)
```

Nested Dictionary Access Explained

employee is a dictionary where one of its values — "skills" — is itself another dictionary. This means reaching a value inside skills requires chaining two key lookups, one after another:

employee["skills"] → selects the outer key "skills", returning the entire nested dictionary, e.g. {"language": "Python", "database": "MySQL", "cloud": "AWS"}
employee["skills"]["language"] → adds a second key lookup on top of that result, reaching into the nested dictionary to pull out just the "language" value, e.g. "Python"

So the first key selects which dictionary to look inside, and the second key selects which field within that inner dictionary. employee["skills"]["language"] reads as: "go to the skills dictionary, then get the value stored under language." The same pattern applies to employee["skills"]["database"] and employee["skills"]["cloud"].

Updating follows the same logic — employee["skills"]["framework"] = "Django" navigates into the nested skills dictionary and adds a brand-new key there, without touching name, department, or salary at the outer level. This is different from employee["experience"] = 3, which adds a new key directly at the outer level of employee, not inside skills.
