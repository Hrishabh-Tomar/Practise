```python
students = [
    ["Rahul", 21, "Python"],
    ["Priya", 22, "Data Science"],
    ["Aman", 20, "Machine Learning"]
]

# Print the complete list
print("Complete list:", students)

# Print Rahul's name
print("Rahul's name:", students[0][0])

# Print Priya's age
print("Priya's age:", students[1][1])

# Print Aman's course
print("Aman's course:", students[2][2])

# Print the complete record of Priya
print("Priya's complete record:", students[1])

# Change Rahul's course to "AI"
students[0][2] = "AI"

# Add another student record manually
students.append(["Sneha", 23, "Web Development"])

# Print the updated nested list
print("\nUpdated nested list:", students)
```

### How Nested-List Indexing Works

`students` is a **list of lists** — the outer list contains three inner lists, and each inner list holds one student's data (`[name, age, course]`).

Because of this two-level structure, accessing a value requires **two indices, one for each level**:

- **First index → selects which inner list (which student)**
  `students[0]` → `["Rahul", 21, "Python"]` (the *entire* record for the first student)
  `students[1]` → `["Priya", 22, "Data Science"]` (the second student's full record)

- **Second index → selects a specific field within that inner list**
  `students[0][0]` → `"Rahul"` (1st item of the 1st inner list → the name)
  `students[1][1]` → `22` (2nd item of the 2nd inner list → the age)
  `students[2][2]` → `"Machine Learning"` (3rd item of the 3rd inner list → the course)

So `students[i][j]` reads as: *"go to student `i`, then get field `j` from that student's record."* This is why `students[1]` alone gives you Priya's whole record, but you need the second index (`students[1][1]`) to drill down into just her age.

**Updating** follows the same logic: `students[0][2] = "AI"` navigates to the first student's inner list, then overwrites index `2` (the course field) — it doesn't touch `name` or `age` at all, since those are different indices within the same inner list.

**Appending**, by contrast, works only at the *outer* level: `students.append([...])` adds a brand-new inner list as a new element of the outer list — it doesn't reach inside any existing student record.
