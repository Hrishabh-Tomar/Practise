```python
laptop = {
    "brand": "Dell",
    "model": "XPS 15",
    "price": 120000,
    "ram": "16GB",
    "storage": "512GB SSD",
    "available": True
}

# Display the product
print("Laptop details:", laptop)

# Print brand
print("\nBrand:", laptop["brand"])

# Print model
print("Model:", laptop["model"])

# Print price
print("Price:", laptop["price"])

# Change price
laptop["price"] = 115000
print("\nAfter changing price:", laptop["price"])

# Add processor information
laptop["processor"] = "Intel Core i7 13th Gen"
print("After adding processor:", laptop["processor"])

# Add GPU information
laptop["gpu"] = "NVIDIA RTX 4060"
print("After adding GPU:", laptop["gpu"])

# Change RAM to "32GB"
laptop["ram"] = "32GB"
print("After changing RAM:", laptop["ram"])

# Remove "available"
del laptop["available"]
print("\nAfter removing 'available':", laptop)

# Print all keys
print("\nAll keys:", laptop.keys())

# Print all values
print("All values:", laptop.values())

# Print all items
print("All items:", laptop.items())

# ------------------------------------------------------------
# Second product dictionary: mobile phone
# ------------------------------------------------------------
mobile_phone = {
    "brand": "Samsung",
    "model": "Galaxy S24 Ultra",
    "price": 105000,
    "ram": "12GB",
    "storage": "256GB",
    "battery": "5000mAh",
    "available": True
}

print("\nMobile phone details:", mobile_phone)
print("Brand:", mobile_phone["brand"])
print("Model:", mobile_phone["model"])
print("Price:", mobile_phone["price"])
```
