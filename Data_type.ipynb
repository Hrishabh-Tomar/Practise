```python
# ---- int ----
age = 21
total_marks = 450

# ---- float ----
price = 99.99
temperature = 36.6

# ---- str ----
name = "Priya"
course = "Data Science"

# ---- bool ----
is_passed = True
is_absent = False

# ---- list ----
fruits = ["apple", "banana", "mango"]
scores = [85, 90, 78, 92]

# ---- tuple ----
coordinates = (12.9716, 77.5946)
rgb_color = (255, 0, 0)

# ---- set ----
unique_numbers = {1, 2, 3, 4}
skills = {"Python", "SQL", "Excel"}

# ---- dict ----
student_info = {"name": "Priya", "age": 21, "course": "Data Science"}

variables = {
    "age": age, "total_marks": total_marks,
    "price": price, "temperature": temperature,
    "name": name, "course": course,
    "is_passed": is_passed, "is_absent": is_absent,
    "fruits": fruits, "scores": scores,
    "coordinates": coordinates, "rgb_color": rgb_color,
    "unique_numbers": unique_numbers, "skills": skills,
    "student_info": student_info,
}

for var_name, value in variables.items():
    print(f"{var_name} = {value}  -->  type: {type(value)}")

```

```python
conv1 = int("100")
print(f'int("100") = {conv1}   type: {type(conv1)}')

conv2 = float("45.67")
print(f'float("45.67") = {conv2}   type: {type(conv2)}')

conv3 = str(500)
print(f'str(500) = "{conv3}"   type: {type(conv3)}')

conv4 = bool(1)
print(f'bool(1) = {conv4}   type: {type(conv4)}')

conv5 = list((1, 2, 3))
print(f'list((1, 2, 3)) = {conv5}   type: {type(conv5)}')

conv6 = tuple([1, 2, 3])
print(f'tuple([1, 2, 3]) = {conv6}   type: {type(conv6)}')

conv7 = set([1, 2, 2, 3])
print(f'set([1, 2, 2, 3]) = {conv7}   type: {type(conv7)}')
```
