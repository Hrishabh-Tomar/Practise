```python
products = [
    {
        "name": "Laptop",
        "price": 70000,
        "brand": "Dell"
    },
    {
        "name": "Phone",
        "price": 40000,
        "brand": "Samsung"
    },
    {
        "name": "Tablet",
        "price": 30000,
        "brand": "Apple"
    }
]

# Print all products
print("All products:", products)

# Print first product
print("\nFirst product:", products[0])

# Print second product's price
print("Second product's price:", products[1]["price"])

# Print third product's brand
print("Third product's brand:", products[2]["brand"])

# Change first product's price
products[0]["price"] = 65000
print("\nAfter changing first product's price:", products[0])

# Add "rating" to the second product
products[1]["rating"] = 4.5
print("After adding rating to second product:", products[1])

# Add another product manually
products.append({
    "name": "Smartwatch",
    "price": 15000,
    "brand": "OnePlus"
})

# Print the final dataset
print("\nFinal dataset:", products)
```

Concept Explanation: Mini E-commerce Dataset

1. Data structure — list of dictionaries
The dataset is modeled as a list where each element is a dict representing one product. This is a common real-world pattern: a list gives you an ordered collection of records, and a dictionary lets each record store labeled fields (name, price, brand) instead of relying on position alone, like a row in a table with named columns.

2. Accessing elements — indexing

products[0] uses list indexing to retrieve the first product (a whole dictionary).
products[1]["price"] chains list indexing + dictionary key lookup — first get the second dict, then pull its "price" value.
products[2]["brand"] follows the same pattern for the third product.

This shows how nested data structures require nested access: index into the list to get a dict, then key into the dict to get a field.

3. Mutability — updating existing data

products[0]["price"] = 65000 demonstrates that dictionaries are mutable: you can reassign a value for an existing key without rebuilding the whole structure.
products[1]["rating"] = 4.5 demonstrates that you can also add a new key to an existing dictionary on the fly — dictionaries aren't fixed-schema, so different products can end up with different sets of fields.

4. Growing the dataset — .append()
products.append({...}) adds a brand-new dictionary as a new element at the end of the list, showing how the overall dataset (not just one record) grows dynamically.
